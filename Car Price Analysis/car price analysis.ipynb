<h1>Car Price Analysis </h1>

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

<h3>Loading Dataset and Viewing First 10 Rows</h3>

```python
df=pd.read_csv("car_price_prediction_.csv")
print(df.head(10))
```

<h3>Finding Duplicate Rows in Data</h3>

```python
df.duplicated().sum()
```

<h3>Finding Null Values in Each Column</h3>

```python
df.isnull().sum()
```

<h3>Calculating Average Car Price</h3>

```python
avg_price = df["Price"].mean()
print(avg_price)
```

<h3>Finding the Highest Priced Car</h3>

```python
max_price=df["Price"].max()
max_price
expensive_car=df[df["Price"]==max_price]
expensive_car
    
```

<h3>Finding the lowesr Priced Car</h3>

```python
min_Price=df["Price"].min()
min_Price
chepest_car=df[df["Price"]==min_Price]
chepest_car
```

<h3>Finding Most Expensive Brand on Average</h3>

```python
avg_price_by_brand = df.groupby("Brand")["Price"].mean()
print(avg_price_by_brand.idxmax())
print(avg_price_by_brand.max())
```

<h3>Bar Chart of Average Car Price per Brand</h3>

```python
plt.bar(avg_price_by_brand.index,avg_price_by_brand.values,
         color="Red",
         edgecolor="black")
plt.xlabel("car_name")
plt.ylabel("avg_price")
plt.title("Average price by brand")
plt.show()
```

<h3>Finding Highest Average Price by Fuel Type</h3>

```python
high_avg_Price=df.groupby("Fuel Type")["Price"].mean()
high_avg_Price
print(high_avg_Price.idxmax())
print(high_avg_Price.max())
```

<h3>Bar Chart of Average Price by Fuel Type</h3>

```python
plt.bar(high_avg_Price.index,
         high_avg_Price.values,
        )
plt.xlabel("Fuel Type")
plt.ylabel("avg_price")
plt.title("Avegage price by Fuel Type")
plt.show()
```

<h3>Finding Year with Highest Average Car Price</h3>

```python
new_car_high_price=df.groupby('Year')["Price"].mean().sort_index()
print(new_car_high_price.idxmax())
print(new_car_high_price.max())
```

<h3>Line Chart of Average Price by Year</h3>

```python
plt.plot(new_car_high_price.index,
         new_car_high_price.values,
         marker="o"
        )
plt.xlabel("Year")
plt.ylabel("average_price")
plt.title("Average_Price by each year")
plt.show()
```

<h3>Finding Highest Average Price by Car Condition</h3>

```python
avg_condition_price=df.groupby("Condition")["Price"].mean()
print(avg_condition_price.idxmax())
print(avg_condition_price.max())
```

<h3>Pie Chart of Car Condition Distribution</h3>

```python
plt.pie(avg_condition_price.values,
        labels=avg_condition_price.index,
        autopct="%1.1f%%"
)
plt.title("condition  distributon")
plt.show()
```

<h3>Finding Brand with Highest Number of Cars</h3>

```python
brand_high_cars=df.groupby("Brand")["Car ID"].count()
print(brand_high_cars.idxmax())
print(brand_high_cars.max())
```

<h3>Finding Fuel Type with Highest Average Mileage</h3>

<h3>Finding Cheapest Brands Based on Average Price</h3>

```python
bottom_5_brand=df.groupby("Brand")["Price"].mean().sort_values().head(5)
print(bottom_5_brand.idxmax())
print(bottom_5_brand.max())
```

<h3>Finding Brands with Highest Price per Mileage</h3>

```python
df["Price_per_Mileage"]=df["Price"]/df["Mileage"]
df["Price_per_Mileage"]
brand_value = df.groupby("Brand")["Price_per_Mileage"].mean()
top5 = brand_value.sort_values(ascending=False).head(5)
print(top5)
```

<h3>Bar Chart of Highest Price per Mileage Brands</h3>

```python
plt.bar(top5.index, top5.values)
plt.title("Top 5 Expensive Brands per Mileage")
plt.xlabel("Brand")
plt.ylabel("Price per Mileage")
plt.show()
```

<h3>Finding Brands with Highest and Lowest Price Variation</h3>

```python
standard=df.groupby("Brand")["Price"].std()
top_5=standard.sort_values(ascending=False).head(5)
bottom_5=standard.sort_values().head(5)
print(top_5)
print(bottom_5)
```

<h3>Bar Charts of Brand-wise Price Variation</h3>

```python
plt.bar(top_5.index,top_5.values,
        color="yellow",
        edgecolor="black"
       )
plt.xlabel("Brand")
plt.ylabel("price vartion")
plt.title("top 5  brand by price vartion")
plt.show()

plt.bar(bottom_5.index,bottom_5.values,
        color="Orange",
        edgecolor="Black"
       )

plt.xlabel("Brand")
plt.ylabel("Price vartion")
plt.title("Botoom 5 brand by price vartion")
plt.show()
```

<h3>Comparing Expensive and Cheap Brands by Price per Engine Size</h3>

```python
df["Price_per_engine"] = df["Price"] / df["Engine Size"]

car_price = df.groupby("Brand")["Price_per_engine"].mean()

top_5 = car_price.sort_values(ascending=False).head(5)
bottom_5 = car_price.sort_values().head(5)
print(top_5)
print(bottom_5)
combined=pd.concat([top_5,bottom_5])
combined
```

<h3>Comparison of Overpriced vs Value Brands Using Price per Engine</h3>

```python
plt.bar(combined.index,combined.values,
        color=["red"]*5 + ["green"]*5
       )
plt.xlabel("Brand")
plt.ylabel("Price per Engine Size")
plt.title("Overpriced vs Value Brands (Price per Engine)")
plt.xticks(rotation=45)
plt.show()
        
        
```
